```python
import pickle
import matplotlib.pyplot as plt

data = pickle.load(open("patch_size_data.pkl", "rb"))
df_vessmap = data["VessMAP"]
df_drive = data["DRIVE"]
```

```python
from matplotlib.ticker import ScalarFormatter

fig, ax = plt.subplots(1, 1, figsize=(4, 3))
ax.plot(df_vessmap.index, df_vessmap["dice/F1"], '-o', label="VessMAP")
ax.plot(df_drive.index, df_drive["dice/F1"], '-o', label="DRIVE")

ax.set_xscale('log', base=2)
xlim = ax.get_xlim()
ax.hlines(df_vessmap["dice/F1"].iloc[-1], xlim[0], xlim[1], colors='gray', linestyles="--", label="Full sample")
ax.hlines(df_drive["dice/F1"].iloc[-1], xlim[0], xlim[1], colors='gray', linestyles="--")
ax.set_xlim(xlim)

ax.set_xticks(df_drive.index)
ax.xaxis.set_major_formatter(ScalarFormatter())
ax.set_xlabel("Patch size")
ax.set_ylabel("Dice")
ax.legend()
fig.tight_layout()

fig.savefig("dice_vs_patch_size.pdf", bbox_inches='tight')
```
